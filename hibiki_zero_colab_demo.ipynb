# Hibiki-Zero (`kyutai/hibiki-zero-3b-pytorch-bf16`) — Hot-Reloadable Colab Runner

This thin Colab runner keeps the **3B model weights loaded in GPU memory** (`bundle`) while pulling code updates from GitHub via `!git pull` + `importlib.reload()`.

### Workflow
1. **Cell 1:** Clone the repo & install dependencies (run once).
2. **Cell 2:** Load `kyutai/hibiki-zero-3b-pytorch-bf16` & `Mimi` into GPU VRAM (`bundle`, run once).
3. **Cell 3 (Hot-Reload & Launch UI):** Run this cell anytime new code is pushed to GitHub — it pulls the latest `.py` files, reloads `hibiki_demo.engine` and `hibiki_demo.ui`, and relaunches the Gradio demo in ~2 seconds **without reloading the 3B model**.

```python
# @title 1. Clone Repository & Install Dependencies (Run Once)
import os
import sys

# Replace with your GitHub repository URL:
REPO_URL = "https://github.com/YOUR_USERNAME/hibiki-zero-demo.git"  # @param {type:"string"}
REPO_DIR = "/content/hibiki-zero-demo"

if not os.path.exists(REPO_DIR):
    !git clone {REPO_URL} {REPO_DIR}
else:
    !git -C {REPO_DIR} pull

if REPO_DIR not in sys.path:
    sys.path.insert(0, REPO_DIR)

!pip install -q -U -r {REPO_DIR}/requirements.txt
```

```python
# @title 2. Load Hibiki-Zero 3B & Mimi into GPU Memory (Run Once)
from hibiki_demo import engine

bundle = engine.load_models("kyutai/hibiki-zero-3b-pytorch-bf16")
```

```python
# @title 3. Pull Latest Code, Hot-Reload Modules & Launch Gradio Demo
# Re-run this cell whenever we push code updates from your local machine!
!git -C /content/hibiki-zero-demo pull

import importlib
from hibiki_demo import engine, ui

importlib.reload(engine)
importlib.reload(ui)

ui.launch_demo(bundle, share=True, inline=True)
```
